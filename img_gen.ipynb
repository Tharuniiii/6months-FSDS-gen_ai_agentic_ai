```python
import tensorflow as tf
```

```python
from tensorflow.keras.preprocessing.image import ImageDataGenerator
from tensorflow.keras.preprocessing.image import load_img, img_to_array
```

```python
# create an img fata generator with augmentations
datagen = ImageDataGenerator(
    rotation_range=40,
    width_shift_range=0.4,
    height_shift_range=0.4,
    shear_range=0.4,
    zoom_range=0.4,
    horizontal_flip=True,
    rescale=None,
    fill_mode='constant' #wrap,constant,nearest,reflect
)
```

```python
# load image from path 
img = load_img(r"C:\Users\Tharuni\Desktop\NIT\oct month\10th_data augumentation\Gemini_Generated_Image_11a8c511a8c511a8.png")
```

```python
img
```

```python
x  = img_to_array(img)
x = x.reshape((1,) + x.shape) # reshaping imges height,width,channels
# generates aug images and saves to given directory
i = 0
for batch in datagen.flow(x, batch_size=1,
                          save_to_dir=r"C:\Users\Tharuni\Desktop\NIT\oct month\10th_data augumentation\Data augmentation",
                         save_prefix='car', save_format='jpeg'):
    i +=1
    if i > 30:
        break
```
