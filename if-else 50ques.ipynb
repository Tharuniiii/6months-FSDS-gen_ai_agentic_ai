Write a program that checks if the input number is a prime
number.

```python
a = int(input('enter 1st num:'))
if a >1:
    for i in range(2,a):
        if a % i == 0:
            print('not prime')
            break
        else:
            print('prime')
else:
    print('not prime')
```

Write a program to count the number of consonants in a string
entered by the user.

```python
text = input("Enter a string: ")
consonants = "bcdfghjklmnpqrstvwxyz"
count = sum(1 for char in text.lower() if char in consonants)
print("Number of consonants:", count)
```

How do you check if a string entered by the user contains any
punctuation?

```python
import string
text = input("Enter a string: ")
if any(char in string.punctuation for char in text):
    print("Contains punctuation")
else:
    print("No punctuation")
```
