```python
nlp = spacy.load('en_core_web_sm')
```

```python
text
```

```python
doc = nlp(text)
doc # we need to convert all the text into documents compulosry
```

```python
# finding the important sentence from para by applying para

tokens = [token.text for token in doc]
print(tokens) # we have created tokens from the para
```

```python
doc
```

```python
# calculating the frequency of each and every tokens in para
word_frequencies = {}

for word in doc:
    if word.text.lower() not in stopwords:
        if word.text.lower() not in punctuation:
            if word.text not in word_frequencies.keys():
                word_frequencies[word.text] = 1
            else:
                word_frequencies[word.text] += 1
```

```python
word_frequencies
```

```python
max_frequency = max(word_frequencies.values())
max_frequency
```

```python
# weighted frequency cal- no.of frequency of word / max frequency of word

for word in word_frequencies.keys():
    word_frequencies[word] = word_frequencies[word]/max_frequency 
```

```python
word_frequencies
```

```python
# calculation the sentences score
sentence_tokens = [sent for sent in doc.sents]
sentence_tokens
```

```python
len(sentence_tokens)
```

```python
sentence_tokens
```

```python
# cal sent score
sentence_score = {}
for sent in sentence_tokens:
    for word in sent:
        if word.text.lower() in word_frequencies:
            if sent not in sentence_score.keys():
                sentence_score[sent] = word_frequencies[word.text.lower()]
            else:
                sentence_score[sent] = word_frequencies[word.text.lower()]
```

```python
sentence_score
```

```python
# 30% sentence with max score
from heapq import nlargest # heapq is dsa it shorts the elements from ascending order
```

```python
select_length = int(len(sentence_tokens)*0.4)
select_length
```

```python
# selecting the max 40% sent out od all
summary = nlargest(select_length,sentence_score,key = sentence_score.get)
```

```python
summary
```

```python
# to combine top 3 sentences
final_summary = [word.text for word in summary]
final_summary
```
