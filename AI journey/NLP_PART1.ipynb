# NLU Stemming

```python
from nltk.stem import PorterStemmer # removes suffixes doesnt strip some words
pst = PorterStemmer()
```

```python
pst.stem('affection')
```

```python
pst.stem('stemming')
```

```python
pst.stem('maximum')
```

```python
words_to_stem = ['give','giving','given','gave']
for words in words_to_stem:
    print(words+':' + pst.stem(words))
```

```python
words_to_stem = ['give','giving','given','gave','tharuni','wholesum','taking']
for words in words_to_stem:
    print(words + ':' + pst.stem(words))
```

```python
from nltk.stem import LancasterStemmer # removes aggresive then porterstem mostly suffix not prefix
lst = LancasterStemmer()

for words in words_to_stem:
    print(words+':' + lst.stem(words))
```

```python
from nltk.stem import SnowballStemmer # acts like porterstemmer when mention english
sbst = SnowballStemmer('english')
for words in words_to_stem:
    print(words+':' + sbst.stem(words))
```

```python
stemmer = SnowballStemmer("french") # supports many languages to strip in own language
stemmer.stem("Merci")
```

```python
from nltk.stem import wordnet
from nltk.stem import WordNetLemmatizer # strip Uses vocabulary + morphological analysis + sometimes POS tagging.
word_lem = WordNetLemmatizer()
```

```python
words_to_stem
```

```python
for words in words_to_stem:
    print(words+':' + word_lem.lemmatize(words))
```
