```python
class computer():
    def __init__(): # dont forget to mention self 
        print('i7','16gb')
c1 = computer()
```

```python
class computer():
    def __init__(self): # dont forget to mention self 
        print('i7','16gb')
c1 = computer()
```

```python
class computer():
    def __init__(self): # special method
        print('i7','16gb')
c1 = computer()
c2 = computer()
c3 = computer()
```

```python
class computer():
    def config(self): # user defined method
        print('i7','16gb')
c1 = computer()
c2 = computer()
c3 = computer()

c1.config()
c2.config()
```

```python
class computer():
    def __init__(self):
        print('in init')
    def config():
        print('i5','64')
com1 = computer() # here system takes only special method first and also config method didnt called
```

```python
class computer():
    def __init__(self):
        print('in init')
    def config():
        print('i5','64')
com1 = computer()
computer.config()
```

```python
class computer():
    def __init__(self):
        print('in init')
    def config(self):
        print('i5','64')
com1 = computer()

com1.__init__()
com1.config() # output 2 times means by default init and again called 2nd time init and config
```
