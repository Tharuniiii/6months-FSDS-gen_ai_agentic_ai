```python
# Method overloading
class demo:
    def add(self,a,b):
        return a + b
d = demo()
print(d.add(2,3))
```

```python
class demo():
    def add(self,a,b):
        return a+b
    def add(self,c,d,e):
        return a + b
d = demo()
print(d.add(2,3)) # same method overloading with diff params
```

```python
class demo:
    def add(self,a,b,c=0,d=0):#use default arg with same positional args
        return a+b+c+d
d = demo()
print(d.add(2,3))
print(d.add(2,3,4))
print(d.add(2,3,4,5)) 
```

```python
class demo:
    def add(self,*args):
        total = 0
        for i in args:
            total = total + i
        return total # using loops we can calculate mul values
d = demo()
print(d.add(2,3))
print(d.add(2,3,4,5))
print(d.add(2,3,4,5,6,7,8))
```

```python
# to avoid infinite loop write exception 
class topten:
    def __init__(self):
        self.num = 1
    def __iter__(self):
        return self
    def __next__(self):
        if self.num <=10:
            val = self.num
            self.num = 1
            return val
        else:
            raise StopIteration 
values = topten
for i in values:
    print(i)
```
