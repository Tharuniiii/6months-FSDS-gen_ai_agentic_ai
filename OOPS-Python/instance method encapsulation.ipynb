# Instance method

```python
class student():
    school = 'nit' # class variables

    def __init__(self,m1,m2,m3): # speacial method
        self.m1 = m1
        self.m2 = m2
        self.m3 = m3
    def avg(self): # instance method 
        return(self.m1+self.m2+self.m3)/3 # instance variables
s1 = student(10,20,30)
s2 = student(5,10,15)

print(s1.avg())
print(s2.avg())
```

# Getter and setter

```python
class student():
    school = 'nit'
    def __init__(self,m1,m2,m3): # special method with instance varibales
        self.m1 = m1
        self.m2 = m2
        self.m3 = m3
    def avg(self): # instance method accesing instance variables
        return(self.m1+self.m2+self.m3)/3

    def get_m1(self): # get the value
        return self.m1
    def set_m1(self,value): # set the value
        self.m1 = value
s1 = student(10,20,30)
s2 = student(5,15,10)

print(s1.avg())
print(s2.avg())
```

 # class method

```python
class student():
    school = 'nit' 
    def __init__(self,m1,m2,m3):
        self.m1 = m1
        self.m2 = m2
        self.m3 = m3
    def avg(self):
        return(self.m1+self.m2+self.m3)/3
    def info(cls): # class method which mandetory to mention cls
        return cls.school # accessing class variable
s1 = student(10,20,30)
s2 = student(15,5,10)

print(s1.avg()) # instance method
print(s2.avg()) 
print(s1.info()) # class method
print(s2.info())

print(s1.school) # class var
print(s1.m1) # instance var
```

# Static method

```python
class student():
    school = 'nit'
    def __init__(self,m1,m2):
        self.m1 = m1
        self.m2 = m2
    def avg(self):
        return(self.m1+self.m2)/2
    @classmethod
    def getschool(cls):
        return cls.school
    @staticmethod # doesnt affect or depend on objects and variables
    def info():
        print('this is student class of static which doesnt access any var or method')
s1 = student(50,50)
s2 = student(10,20)

print(s1.avg(),s2.avg())
print(student.getschool())
student.info()
```
