# self as argument and self as object

```python
class computer():
    def __init__(self,cpu,ram): # here self acts default arg
        self.cpu =  cpu# self as obj
        self.ram = ram
    def config(self):
        print('config is:', self.cpu,self.ram)
com1 = computer('i5',8)
com2 = computer('i7',16)

com1.config()
com2.config()
```

```python
class computer():
    def __init__(self,cpu,ram,price,company):
        self.cpu = cpu
        self.ram = ram
        self.price = price
        self.company = company
    def config(self):
        print('config is -',self.cpu,self.ram,self.price,self.company)

com1 = computer('i5',8,50000,'hp')
com2 = computer('i7',9,64000,'azure')

computer.config(com1)
computer.config(com2)
```

# Encapsulation

```python
class car():
    def __init__(self):
        self.mil = 10
        self.com = 'bmw'
c1 = car()
c2 = car()

c1.mil = 8
c2.mil = 20
c1.com = 'maruti'

print(c1.mil,c1.com)
print(c2.mil,c2.com)
```
