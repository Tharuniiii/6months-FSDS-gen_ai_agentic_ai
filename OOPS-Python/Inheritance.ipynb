# multi-level 

```python
class A():
    def feature1(self):
        print('feature1 from A')
    def feature2(self):
        print('feature2 from A')
class B(): # no inheriting 
    def feature3(self):
        print('feature3 from B')
    def feature4(self):
        print('feature4 from B')
class c(A,B): # here acquiring a and b class properties
    def feature5(self):
        print('feature5 from c')

a1 = A()
b1 = B()
c1 = c()
c1.feature1()
c1.feature2()
c1.feature3()
c1.feature4()
c1.feature5()
```

# Hybrid level-multilevel + multiple

```python
class A():
    def feature1(self):
        print("A class")
class B(A):
    def feature2(self):
        print("B class")
class C(A):
    def feature3(self):
        print("C class")
class D(B,C):
    def feature4(self):
        print("D class")

d1 = D()
d1.feature1()
d1.feature2()
d1.feature3()
d1.feature4()
```

# Super() Function

```python
class A():
    def __init__(self):
        print('A class init')
    def feature1(self):
        print('A feature1')
class B(A): # inherits A class so if no init in B class no error it will acquire
    def __init__(self):
        super().__init__() # now it will print parent class init
        print('B class init')
    def feature2(self):
        print('B feature2')

b1 = B()
```

```python
# super with multiple
class A():
    def __init__(self):
        print('A class init')
    def feature1(self):
        print('A feature1')
class B(): # inherits A class so if no init in B class no error it will acquire
    def __init__(self):
        print('B class init')
    def feature2(self):
        print('B feature2')

class C(A,B):
    def __init__(self):
        super().__init__()
        print('c class init')

c1 = C() #prints only a and c class init the prblm is MRO
```

```python
class duck():
    def swim(self):
        print('i am duck & i can swim')
    def speaks(self):
        print('quack quack')

class dog(): # different classes but methods should same
    def swim(self):
        print('i am dog & i can swim')
    def speaks(self):
        print('woof woof')


def display(duck): # we define outside of the class
    duck.swim()
    duck.speaks()
    #print('informatation')

d = duck()
display(d)       
```
