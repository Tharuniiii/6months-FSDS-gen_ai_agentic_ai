```python
import matplotlib.pyplot as plt
```

```python

plt.figure(figsize=(8,6))
```

# scatter plot

```python
sns.scatterplot(data=tips,x="total_bill", y="tip", hue="time", size="size", palette="deep")
plt.title("scatterplot of total bill and tip ")
plt.show()
```

# Line Plot

```python
sns.lineplot(data=tips,x="size", y="total_bill", hue="sex", marker="o")
plt.title("lineplot of toal bill vs size ")
plt.show()
```

```python
sns.lineplot(data=tips,x="size", y="total_bill", hue="sex",ci = None, marker="o")
plt.title("lineplot of toal bill vs size ")
plt.show()
```

# Bar plot

```python
sns.barplot(data = tips, x='day', y='total_bill', hue='sex')
plt.title("barplot of bill by day")
plt.show()
```

# Box Plot

```python
tips.columns
```

```python
sns.boxplot(data = tips, x='day', y='tip', hue='smoker', palette= 'Set2' )
plt.title("boxplot of day by tip")
plt.show()
```

# violin plot

```python
sns.violinplot(data = tips, x='day', y='total_bill', hue='time', split =True, palette= 'pastel')
plt.title("violinplot of totalbill by day")
plt.show()
```

# Count Plot

```python
tips.columns
```

```python
sns.countplot(data = tips, x='day', hue='smoker', palette= 'dark')
plt.title("countplot of Days by smoker status")
plt.show()
```

# regression plot

```python
sns.regplot(data = tips, x='total_bill',y='tip',scatter_kws={'s':50},line_kws={'color':'red'})
plt.title("regression plot of totalbill vs tip")
plt.show()
```

# Histrogram plot

```python
sns.histplot(data=tips,x='total_bill', bins = 20, kde=True,color = 'blue')
plt.title("histogram of total bill")
plt.show()
```

# Pair Plot(relationship b/w 2 variables)

```python
sns.pairplot(data=tips,hue='sex',vars=["total_bill","tip","size"],palette='husl')
plt.suptitle("pair plot: numerical variable by gender", y=1.02)
plt.show()
```

# categorical plot

```python
sns.catplot(data=tips, x='day',y='tip',hue='sex',kind='point',palette='bright')
plt.suptitle("catplot(point):Tips by day and gender")
plt.show()
```

# Joint plot

```python
sns.jointplot(data=tips,x='total_bill',y='tip',kind='scatter',hue='smoker',palette='coolwarm')
plt.suptitle("jointplot:Total bill vs tip",y=1.02)
plt.show()
```

# facet Grid

```python
g=sns.FacetGrid(tips,col='time',row='smoker',margin_titles=True).map(sns.histplot,'total_bill',bins=20,kde=True).add_legend()
g
```

# Strip Plot

```python
sns.stripplot(tips,x='day',y='tip',hue='sex',jitter=True,palette='Set1')
plt.title("StripPlot:tips by day and gender")
plt.show()
```

# KDE Plot(kernel density plot)

```python
sns.kdeplot(data=tips,x='total_bill',hue='sex', fill=True, palette='tab10')
plt.title("kde plot:Total bill density by gender")
plt.show()
```
