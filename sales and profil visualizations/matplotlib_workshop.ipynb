```python
import pandas as pd

data = {
    "Month" : ['jan','feb','mar','apr','may','june'],
    "Sales" : [10000,12000,15000,13000,17000,16000],
    "Profit" : [2000,3000,4000,2500,3500,3000]
}
df = pd.DataFrame(data)
print(df)
```

# line plot monthly sales

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(6,3))
plt.plot(df['Month'],df['Sales'],color = 'blue',marker = 'o',linestyle = '-',label = 'Sales')
plt.title('Sales trends over month')
plt.xlabel('Month')
plt.ylabel('Sales ($)')
plt.grid(True)
plt.legend()
plt.show()
```

# barplot month vs profit

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(6,3))
width = 0.3
plt.bar(df['Month'], df['Sales'], width=width, color = 'skyblue', label = 'Sales')
plt.bar(df['Month'], df['Profit'], width=width, color = 'green', label = 'Sales', bottom = df['Sales'])
plt.title('Sales and Profit Trends By Month')
plt.xlabel('Month')
plt.ylabel('Amount ($)')
plt.grid(True)
plt.legend()
plt.tight_layout()
plt.show()
```

# pie chart profit vs month

```python
plt.figure(figsize=(10,5))
plt.pie(df['Profit'], labels=df['Month'], autopct='%1.2f%%', startangle=140,colors = plt.cm.Paired.colors) #autopct is format string parameter
plt.title('Profit by Month')
plt.show()
```

# scatter plot

```python
plt.figure(figsize=(8, 5))
plt.scatter(df['Sales'], df['Profit'], color='blue',s = 100,edgecolors='black',marker='o')
plt.title('Sales vs Profit Scatter Plot')
plt.xlabel('Sales')
plt.ylabel('Profit')
plt.tight_layout()
plt.grid(True)
plt.show()
```

# histrogram plot

```python
plt.figure(figsize=(8, 4))
plt.hist(df['Sales'],bins=5,color='purple',edgecolor='black'),
plt.title('Sales Distribution'),
plt.xlabel('Sales'),
plt.ylabel('Frequency'),
plt.tight_layout(),
plt.show()
```

# box plot

```python
plt.figure(figsize=(8, 4))
plt.boxplot(df['Profit'],vert=False,patch_artist=True, boxprops=dict(facecolor='lightgreen')),
plt.title('sales box plot'),
plt.ylabel('Profit'),
plt.tight_layout(),
plt.show()
```

```python
import gradio as gr
import  pandas as pd
import matplotlib.pyplot as plt

data = {
    "Month" : ['jan','feb','mar','apr','may','june'],
    "Sales" : [10000,12000,15000,13000,17000,16000],
    "Profit" : [2000,3000,4000,2500,3500,3000]
}
df = pd.DataFrame(data)
def generate_plot(plot_type):
    fig, ax = plt.subplots(figsize=(8, 5))

    if plot_type == 'Line Plot':
       ax.plot(df['Month'],df['Sales'],color = 'blue',marker = 'o',linestyle = '-',label = 'Sales')
       ax.set_title('Sales trends over month')
       ax.set_xlabel('Month')
       ax.set_ylabel('Sales ($)')
       ax.grid(True)
       ax.legend()
    elif plot_type == 'stacked bar chart':
      fig.set_size_inches(10, 6)
      width = 0.3
      ax.bar(df['Month'], df['Sales'], width=width, color = 'skyblue', label = 'Sales')
      ax.bar(df['Month'], df['Profit'], width=width, color = 'green', label = 'Profit', bottom = df['Sales'])
      ax.set_title('Sales and Profit Trends By Month')
      ax.set_xlabel('Month')
      ax.set_ylabel('Amount ($)')
      ax.grid(True)
      ax.legend()
      fig.tight_layout()
    elif plot_type == 'pie chart':
     fig.set_size_inches(7, 7)
     ax.pie(df['Profit'], labels=df['Month'], autopct='%1.2f%%', startangle=140,colors = plt.cm.Paired.colors) #autopct is format string parameter
     ax.set_title('Profit by Month')
    elif plot_type == 'scatter plot':
     ax.scatter(df['Sales'], df['Profit'], color='blue',s = 100,edgecolors='black',marker='o')
     ax.set_title('Sales vs Profit Scatter Plot')
     ax.set_xlabel('Sales')
     ax.set_ylabel('Profit')
     fig.tight_layout()
     ax.grid(True)
    elif plot_type == 'histrogram plot':
     ax.hist(df['Sales'],bins=5,color='purple',edgecolor='black'),
     ax.set_title('Sales Distribution'),
     ax.set_xlabel('Sales'),
     ax.set_ylabel('Frequency'),
    elif plot_type == 'box plot':
     ax.boxplot(df['Profit'],vert=False,patch_artist=True, boxprops=dict(facecolor='lightgreen')),
     ax.set_title('Profit box plot'),
     ax.set_xlabel('Profit'),
    fig.tight_layout()
    return fig
# gradio ui

demo = gr.Interface(
    fn =generate_plot,
    inputs=gr.Radio(['Line Plot','stacked bar chart','pie chart','scatter plot','histrogram plot','box plot']),
    outputs=gr.Plot(label="sales data visulaization"),
    title='Sales and profit visual insight',
    description="choose the type to visualize the data"
    )
demo.launch()
```
